# Daisy Z-stage - Quickstart Guide

This notebook will teach you how to connect your **Daisy Z-stage**, home it to its reference position, and move it to a target position.

## Home the stage

- Always call `stage.home()` before first use. This moves the stage to Z = 0 using the end-stop for a reliable reference position.

## Move the stage

- Use `stage.run(z, wait=False)` to move the stage to a position with arguments:
>    `z` — target position in mm (clamped to 0–45 mm) \
>    `wait` — if `True`, waits for the move to complete before running the next line
- Use `stage.check_position_mm()` to return the current position of the Z-stage in mm

Running the Z-stage to a value above 45 mm will result in running the stage at 45 mm.

## Import
The Daisy Python Library is already loaded into the Daisy Controller Software, all you have to do is import any external libraries you need. Here we are importing random to select different distances within the Z-stage range using `random.uniform(0,45)`.

```python
import random
```

## Print instruments found
Print the instruments found to easily verify the index of the Daisy you will be using.

```python
print("Instruments found:")
for position, (code, inst) in enumerate(zip(self.my_interface.instruments_list[0], self.my_interface.groups[0].inst)):
    print(f"  [{position}]  {code} — {inst.name}")
```

## Set your instrument variable name
Change `INST_INDEX` to the position of the Daisy (Z-stage) you want to use.

```python
# Set which instrument you are using
INST_INDEX = 0   # 0 if first instrument or change to match your Z-stage's position

stage = self.my_interface.groups[0].inst[INST_INDEX]
print(f"Using: {stage.name}")
```

## Home the stage

Always home the stage before first use. This moves it to Z = 0 using the end-stop as a reference so all subsequent positions are accurate.

```python
# Home your Z-stage
stage.home()
print("Homing complete. Stage is at Z = 0 mm.")
```

```python
# Target position in mm (0–45 mm)
TARGET_Z_MM = 20.0

stage.run(TARGET_Z_MM, wait=True)
print(f"Stage moved to Z = {stage.check_position_mm()} mm")
```

## Example moves

**To stop the stage while it is moving:** Use the Jupyter stop button (■) in the toolbar (or press `I` twice to interrupt the kernel). Then run `stage.stop()`. If you stop the stage while it is moving, you don't have to rehome the stage.

```python
# Move to 20 mm, wait until arrived, then move back to 0
stage.run(20.0, wait=True)
print("At 20 mm")
stage.run(0.0, wait=True)
print("Back at 0 mm")
```

```python
# Start a move without waiting, then do other work while it travels
stage.run(35.0, wait=False)
print("Stage is moving. This line prints immediately.")
```

```python
# Randomly select a target position from 0 - 45, 3 times, move to position then return home
for i in range(3):
    z = random.uniform(0, 45)
    print(f"Moving to {z:.1f} mm")
    stage.run(z, wait=True)
    stage.home(wait=True)
```

```python
# Run this cell if you need to stop the stage while it is moving
stage.stop()
```
